## 0 · El problema y el algoritmo

Una persona adquiere **10 acciones a USD 150** cada una. Posteriormente, el precio hipotético es **USD 165**. Queremos conocer el capital invertido, el valor actual, el resultado monetario y el rendimiento porcentual.

| Entradas | Procesos | Salidas |
|---|---|---|
| Cantidad, precio de compra, precio actual | Multiplicar, restar, dividir | Inversión, valor actual, resultado, rendimiento |

**Fórmulas:**

- Inversión inicial = cantidad × precio de compra.
- Valor actual = cantidad × precio actual.
- Ganancia o pérdida = valor actual − inversión inicial.
- Rendimiento simple (%) = (ganancia o pérdida / inversión inicial) × 100.

**Pseudocódigo:**

```text
INICIO
 LEER cantidad, precio_compra, precio_actual
 VALIDAR cantidad > 0 y precio_compra > 0
 inversion = cantidad * precio_compra
 valor = cantidad * precio_actual
 resultado = valor - inversion
 rendimiento = resultado / inversion * 100
 MOSTRAR inversion, valor, resultado, rendimiento
FIN
```

## 1 · Variables, tipos de datos y operadores

Una **variable** identifica un valor. Los tipos `str`, `int`, `float` y `bool` permiten representar texto, enteros, decimales y valores lógicos. El operador `=` asigna un valor, mientras que `==` compara valores. Los nombres de las variables deben reflejar su significado financiero.

````python
activo = "ACT_A"  # Activo ficticio
cantidad = 10
precio_compra = 150.0
precio_actual = 165.0
precios_hipoteticos = True
inversion = cantidad * precio_compra
valor_actual = cantidad * precio_actual
resultado = valor_actual - inversion
rendimiento_pct = resultado / inversion * 100
print(f"Inversión: USD {inversion:,.2f}")
print(f"Valor actual: USD {valor_actual:,.2f}")
print(f"Resultado: USD {resultado:,.2f}")
print(f"Rendimiento: {rendimiento_pct:.2f}%")
````

**Interpretación:** el capital invertido es USD 1.500 y el valor actual USD 1.650. La ganancia hipotética es USD 150, equivalente al 10 %. El rendimiento no incluye dividendos, comisiones ni impuestos.

## 4 · Ciclos: calcular el resultado conjunto

Un ciclo `for` recorre una colección. Los **acumuladores** suman resultados parciales. El rendimiento agregado del portafolio se obtiene dividiendo la ganancia o pérdida total entre el capital total invertido: no es, en general, el promedio simple de los porcentajes individuales.

````python
portafolio_base = [
    {"ticker": "ACT_A", "cantidad": 10, "compra": 150.0, "actual": 165.0},
    {"ticker": "ACT_B", "cantidad": 5, "compra": 80.0, "actual": 76.0},
    {"ticker": "ACT_C", "cantidad": 12, "compra": 40.0, "actual": 43.0}
]
capital_total = 0
valor_total = 0
for posicion in portafolio_base:
    capital = posicion["cantidad"] * posicion["compra"]
    valor = posicion["cantidad"] * posicion["actual"]
    capital_total += capital
    valor_total += valor
    print(posicion["ticker"], "Resultado USD:", valor - capital)
resultado_total = valor_total - capital_total
rendimiento_total = resultado_total / capital_total * 100
print(f"Capital: USD {capital_total:,.2f}; Valor: USD {valor_total:,.2f}")
print(f"Resultado: USD {resultado_total:,.2f}; Rendimiento: {rendimiento_total:.2f}%")
````

El ciclo `while` repite un bloque mientras una condición sea verdadera. Para evitar repeticiones infinitas, alguna instrucción debe permitir que la condición deje de cumplirse.

````python
intentos = 0
while intentos < 3:
    intentos += 1
    print("Validación simulada", intentos)
````

## 5 · Funciones y validación

Una **función** encapsula un procedimiento. Sus parámetros reciben entradas y `return` devuelve resultados. Validar antes de calcular evita divisiones por cero y cantidades incoherentes.

````python
def analizar_posicion(cantidad, compra, actual):
    if cantidad <= 0 or compra <= 0 or actual < 0:
        raise ValueError("Verifica cantidad y precios.")
    capital = cantidad * compra
    valor = cantidad * actual
    resultado = valor - capital
    return {"capital": capital, "valor": valor,
            "resultado": resultado, "rendimiento_pct": resultado / capital * 100}

prueba = analizar_posicion(10, 150, 165)
print(prueba)
assert prueba["capital"] == 1500
assert prueba["valor"] == 1650
assert prueba["resultado"] == 150
assert abs(prueba["rendimiento_pct"] - 10) < 1e-10
print("Resultados comprobados.")
````

**Interpretación:** `assert` permite contrastar los cálculos con resultados conocidos. Si una condición no se cumple, se produce un error que señala una discrepancia.

## 6 · Integración: primer analizador de portafolios

La función siguiente recorre las posiciones, valida cada una y agrega sus resultados. Separa los cálculos del texto que se muestra en pantalla para facilitar la revisión.

````python
def analizar_portafolio(posiciones):
    if not posiciones:
        raise ValueError("El portafolio está vacío.")
    detalle = []
    capital_total = 0
    valor_total = 0
    for p in posiciones:
        analisis = analizar_posicion(p["cantidad"], p["compra"], p["actual"])
        detalle.append({"ticker": p["ticker"], **analisis})
        capital_total += analisis["capital"]
        valor_total += analisis["valor"]
    return {"detalle": detalle, "capital": capital_total,
            "valor": valor_total, "resultado": valor_total - capital_total,
            "rendimiento_pct": (valor_total - capital_total) / capital_total * 100}

reporte = analizar_portafolio(portafolio_base)
for fila in reporte["detalle"]:
    print(f'{fila["ticker"]}: {fila["rendimiento_pct"]:.2f}%')
print(f'Rendimiento del portafolio: {reporte["rendimiento_pct"]:.2f}%')
assert reporte["capital"] == 2380
assert reporte["valor"] == 2546
assert reporte["resultado"] == 166
print("Verificación superada.")
````

**Interpretación financiera:** ACT_A aporta USD 150; ACT_B registra una pérdida de USD 20 y ACT_C aporta USD 36. El resultado total es USD 166 sobre USD 2.380 invertidos, aproximadamente **6,97 %**. Este porcentaje no se obtiene sumando los rendimientos individuales.

**Limitaciones:** los precios son hipotéticos, las cantidades se mantienen constantes y se omiten comisiones, dividendos, impuestos y flujos adicionales. El análisis compara dos momentos; todavía no utiliza series temporales.
