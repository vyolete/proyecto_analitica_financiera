**Ciclos y funciones:** una función encapsula un cálculo; un ciclo permite aplicarlo a varios productos.

```python
def calcular_factura(precio, unidades, iva=0.19):
    base = precio * unidades
    return {"subtotal": base, "iva": base * iva, "total": base * (1 + iva)}

productos = [("Cuaderno", 12000, 2), ("Lapicero", 3500, 4), ("Carpeta", 7000, 1)]
for nombre, precio, unidades in productos:
    factura = calcular_factura(precio, unidades)
    print(nombre, "→", round(factura["total"], 2))
```

## Etapa 3 · Análisis exploratorio y portafolios
**Equivalencias propuestas:** S09 (entrega mayor 1, estructuras aplicadas a un caso financiero) y S15 (EDA).

**Caso real:** comparar tres activos cotizados mediante datos históricos de Yahoo Finance. Los precios pueden cambiar y la fuente puede estar temporalmente indisponible. No se emiten recomendaciones de inversión.

```python
import yfinance as yf
tickers = ["MSFT", "AAPL", "JPM"]
precios_descargados = yf.download(tickers, period="1y", auto_adjust=True, progress=False)
if precios_descargados.empty:
    raise RuntimeError("No se recibieron datos de Yahoo Finance. Reintenta más tarde.")
precios = precios_descargados["Close"].copy()
if isinstance(precios, pd.Series):
    precios = precios.to_frame(name=tickers[0])
precios = precios.dropna(how="all").ffill().dropna()
if len(precios) < 30:
    raise RuntimeError("No hay suficientes observaciones para el análisis.")
display(precios.head())
print("Filas y columnas:", precios.shape)
```

**Rendimiento simple diario:** \(r_t = P_t/P_{t-1}-1\). Un rendimiento de 0,01 equivale a 1 %.

```python
rendimientos = precios.pct_change().dropna()
resumen = pd.DataFrame({
    "media_diaria": rendimientos.mean(),
    "mediana_diaria": rendimientos.median(),
    "volatilidad_diaria": rendimientos.std(),
    "minimo": rendimientos.min(),
    "maximo": rendimientos.max(),
})
display(resumen.round(5))
precios_normalizados = precios.div(precios.iloc[0]).mul(100)
precios_normalizados.plot(figsize=(10, 4), title="Evolución de precios normalizados (base 100)")
plt.ylabel("Índice (base 100)")
plt.grid(alpha=0.2)
plt.show()
```

## Etapa 4 · Series temporales y riesgo financiero
**Equivalencias propuestas:** S21 (series temporales) y Quiz 3 S22 (visualización). Se confirmará si la evaluación exige además un componente en R/ggplot2.

**Riesgo de un portafolio:** un promedio ponderado de rendimientos y una matriz de covarianzas permiten estimar su variabilidad histórica. No garantizan rendimientos futuros.

```python
pesos = np.array([0.40, 0.35, 0.25])
assert len(pesos) == rendimientos.shape[1], "Revisa la cantidad de activos."
assert np.isclose(pesos.sum(), 1), "Los pesos deben sumar 1."
retorno_portafolio = rendimientos.dot(pesos)
volatilidad_diaria_portafolio = np.sqrt(pesos @ rendimientos.cov().to_numpy() @ pesos)
indice_portafolio = (1 + retorno_portafolio).cumprod()
drawdown = indice_portafolio / indice_portafolio.cummax() - 1
print("Volatilidad diaria histórica:", round(volatilidad_diaria_portafolio, 5))
print("Máxima caída histórica:", round(drawdown.min(), 5))
```
