**¿Qué es un algoritmo?** Es una secuencia ordenada de instrucciones para resolver un problema. Antes de escribir código conviene identificar las entradas (por ejemplo, precio y cantidad), el procesamiento (subtotal, impuesto, descuento) y las salidas (valor total de la factura). **¿Qué es una variable?** Es un nombre que referencia un valor que el programa puede utilizar. En Python, `precio = 25000` permite usar ese valor en operaciones posteriores. Los tipos de datos importan: una cantidad suele ser un entero (`int`), un precio puede ser numérico (`int` o `float`) y el nombre de un producto es texto (`str`).

**Operadores y cálculo de IVA.** Los operadores `+`, `-`, `*` y `/` permiten realizar cálculos. En el ejemplo, el subtotal se obtiene multiplicando precio por cantidad; el IVA se calcula multiplicando el subtotal por una tasa decimal; el total suma ambos valores. Una tasa del 19 % se escribe `0.19`, no `19`. La tasa es un parámetro didáctico y no sustituye la revisión de las reglas tributarias aplicables.

**Qué demostrarás:** que puedes convertir una situación de facturación en un algoritmo y programarlo con variables, decisiones, ciclos y funciones.

### Repeticiones y funciones: `for` y `def`
Un **ciclo `for`** recorre los elementos de una colección. Es útil cuando hay varios productos: evita repetir manualmente el mismo cálculo para cada uno. Una **función**, creada con `def`, agrupa instrucciones que realizan una tarea específica; recibe valores de entrada (*parámetros*) y puede devolver resultados mediante `return`.

En el ejemplo, `calcular_factura` recibe precio, unidades y tasa de IVA; devuelve un diccionario con subtotal, IVA y total. El ciclo aplica la función a cada producto. Esta separación permite reutilizar el cálculo y comprobarlo con distintos valores.

```python
def calcular_factura(precio, unidades, iva=0.19):
    base = precio * unidades
    return {"subtotal": base, "iva": base * iva, "total": base * (1 + iva)}

productos = [("Cuaderno", 12000, 2), ("Lapicero", 3500, 4), ("Carpeta", 7000, 1)]
for nombre, precio, unidades in productos:
    factura = calcular_factura(precio, unidades)
    print(nombre, "→", round(factura["total"], 2))
```

### ¿Cómo se calcula e interpreta un rendimiento?
El **rendimiento simple** mide el cambio relativo del precio entre dos fechas consecutivas:

\[
r_t=\frac{P_t-P_{t-1}}{P_{t-1}}=\frac{P_t}{P_{t-1}}-1
\]

Aquí, \(P_t\) es el precio actual y \(P_{t-1}\) el precio anterior. Si un activo pasa de 100 a 103, su rendimiento es \((103/100)-1=0.03\), equivalente a **3 %**. Si pasa de 100 a 97, es **−3 %**.

Un rendimiento promedio positivo no significa que todos los días fueron positivos. Por eso conviene analizar la distribución completa y su dispersión, no solo una cifra. El cálculo `pct_change()` de pandas aplica esta fórmula a una serie ordenada temporalmente.

```python
rendimientos = precios.pct_change().dropna()
resumen = pd.DataFrame({
    "media_diaria": rendimientos.mean(),
    "mediana_diaria": rendimientos.median(),
    "volatilidad_diaria": rendimientos.std(),
    "minimo": rendimientos.min(),
    "maximo": rendimientos.max(),
})
display(resumen.round(5))
precios_normalizados = precios.div(precios.iloc[0]).mul(100)
precios_normalizados.plot(figsize=(10, 4), title="Evolución de precios normalizados (base 100)")
plt.ylabel("Índice (base 100)")
plt.grid(alpha=0.2)
plt.show()
```

## Etapa 4 · Series temporales y riesgo financiero
**Material de estudio:** notebooks 04 y 05.

**Serie temporal:** conjunto de observaciones registradas en orden cronológico. En finanzas, los precios diarios y los rendimientos forman series temporales; la fecha no es una etiqueta decorativa, sino una dimensión necesaria para interpretar la evolución.

**Media móvil:** promedio calculado sobre una ventana de observaciones recientes. Una media móvil de 20 días resume los últimos 20 registros disponibles y ayuda a visualizar tendencias, pero no predice por sí sola el siguiente precio.

**Volatilidad móvil:** desviación estándar de los rendimientos calculada sobre una ventana que se desplaza. Permite observar cómo cambia la variabilidad histórica; una mayor volatilidad implica oscilaciones más amplias, no necesariamente pérdidas.

**Correlación:** medida de asociación lineal entre dos series; sus valores están entre −1 y 1. Correlación positiva significa que tienden a moverse en la misma dirección; negativa, en direcciones opuestas. No implica causalidad.

**Portafolio:** combinación de activos con pesos que suman 100 %. Su rendimiento depende de los rendimientos individuales y de sus ponderaciones; su riesgo también depende de cómo se mueven conjuntamente los activos.

**Equivalencias propuestas:** S21 y comprobación formativa 3 S22, pendiente de confirmar si exige R/ggplot2.

```python
media_movil_20 = precios.rolling(window=20).mean()
volatilidad_movil_20 = rendimientos.rolling(window=20).std()
fig, ax = plt.subplots(figsize=(10, 4))
activo = tickers[0]
precios[activo].plot(ax=ax, label="Precio")
media_movil_20[activo].plot(ax=ax, label="Media móvil 20 días")
ax.set_title(f"{activo}: precio y media móvil")
ax.legend()
ax.grid(alpha=0.2)
plt.show()
display(rendimientos.corr().round(3))
```

### Riesgo de portafolio: volatilidad y caída máxima
El **rendimiento del portafolio** se obtiene combinando los rendimientos de los activos según sus pesos. En este ejercicio se supone que los pesos se mantienen constantes en cada período (rebalanceo periódico idealizado).

La **covarianza** mide cómo varían conjuntamente dos rendimientos; la **matriz de covarianzas** reúne estas relaciones. La volatilidad del portafolio se estima con:

\[
\sigma_p=\sqrt{\mathbf{w}^{T}\Sigma\mathbf{w}}
\]

donde \(\mathbf{w}\) contiene los pesos y \(\Sigma\) es la matriz de covarianzas. La unidad temporal de la volatilidad corresponde a la frecuencia de los datos: aquí es **diaria**.

El **drawdown** mide cuánto ha caído el valor acumulado respecto de su máximo anterior. La **máxima caída** es el valor más negativo observado en ese indicador. Volatilidad y drawdown describen aspectos distintos del riesgo y no predicen pérdidas futuras.

```python
pesos = np.array([0.40, 0.35, 0.25])
assert len(pesos) == rendimientos.shape[1], "Revisa la cantidad de activos."
assert np.isclose(pesos.sum(), 1), "Los pesos deben sumar 1."
retorno_portafolio = rendimientos.dot(pesos)
volatilidad_diaria_portafolio = np.sqrt(pesos @ rendimientos.cov().to_numpy() @ pesos)
indice_portafolio = (1 + retorno_portafolio).cumprod()
drawdown = indice_portafolio / indice_portafolio.cummax() - 1
print("Volatilidad diaria histórica:", round(volatilidad_diaria_portafolio, 5))
print("Máxima caída histórica:", round(drawdown.min(), 5))
```
