## 4. Ciclos: procesar información repetida

Un ciclo `for` recorre una colección. El **acumulador** registra resultados parciales y permite sumar todas las referencias sin escribir un cálculo independiente para cada una.

````python
subtotal_factura = 0
for item in carrito:
    subtotal_item = item["precio"] * item["cantidad"]
    subtotal_factura += subtotal_item
    print(item["nombre"], f"${subtotal_item:,.0f}")
print("Subtotal factura:", subtotal_factura)
````

Un ciclo `while` repite instrucciones mientras una condición sea verdadera. Es importante actualizar la condición para evitar repeticiones infinitas.

````python
saldo = 30000
abono = 10000
while saldo > 0:
    saldo -= abono
    print("Saldo pendiente:", saldo)
````

## 5. Funciones: organizar cálculos reutilizables

Una función encapsula un procedimiento que puede aplicarse a diferentes datos. Sus **parámetros** reciben valores y `return` devuelve un resultado. Separar el cálculo de su presentación ayuda a comprobarlo y reutilizarlo.

````python
def calcular_total(precio, cantidad, tasa_descuento=0.0, tasa_iva=0.19):
    """Calcula una referencia y devuelve sus indicadores."""
    if precio < 0 or cantidad <= 0:
        raise ValueError("Precio y cantidad deben ser válidos.")
    if not 0 <= tasa_descuento <= 1 or not 0 <= tasa_iva <= 1:
        raise ValueError("Las tasas deben estar entre cero y uno.")
    subtotal = precio * cantidad
    descuento = subtotal * tasa_descuento
    base = subtotal - descuento
    iva = base * tasa_iva
    return {"subtotal": subtotal, "descuento": descuento, "iva": iva, "total": base + iva}
print(calcular_total(8500, 3))
````

## 6. Integración guiada: factura de varias referencias

Integraremos el ciclo y las estructuras de datos en una función. El ejemplo no contempla descuentos por referencia; utiliza el mismo IVA didáctico para todos los productos.

````python
def resumir_factura(items, tasa_iva=0.19):
    subtotal = 0
    for item in items:
        if item["precio"] < 0 or item["cantidad"] <= 0:
            raise ValueError("Precio o cantidad inválidos.")
        subtotal += item["precio"] * item["cantidad"]
    iva = subtotal * tasa_iva
    return {"subtotal": subtotal, "iva": iva, "total": subtotal + iva}

referencia = [
    {"nombre": "Cuaderno", "precio": 8500, "cantidad": 2},
    {"nombre": "Lápiz", "precio": 1500, "cantidad": 4},
    {"nombre": "Carpeta", "precio": 4000, "cantidad": 1}
]
resultado = resumir_factura(referencia)
print(resultado)
assert resultado["subtotal"] == 27000
assert abs(resultado["iva"] - 5130) < 1e-8
assert abs(resultado["total"] - 32130) < 1e-8
print("Verificación superada.")
````

**Interpretación:** el subtotal de $27.000 procede de $17.000 en cuadernos, $6.000 en lápices y $4.000 en carpeta. El IVA del ejemplo es $5.130 y el total $32.130. Las expresiones `assert` permiten verificar que el programa produce esos resultados.
